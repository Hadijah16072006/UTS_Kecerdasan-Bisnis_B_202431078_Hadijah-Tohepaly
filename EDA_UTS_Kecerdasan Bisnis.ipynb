```python
# Hadijah Tohepaly - 202431078
# import library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Hadijah Tohepaly - 202431078
# Load hasil ETL

df_clean = pd.read_csv("bank_cleaned.csv")

print(df_clean.shape)
```

```python
# Hadijah Tohepaly - 202431078
# Statistik deskriptif

df_clean.describe()
```

```python
# Hadijah Tohepaly - 202431078
# Visualisasi 1: Distribusi Status Deposito

plt.figure(figsize=(8,5))

sns.countplot(data=df_clean, x="status_deposito")

plt.title("Distribusi Status Berlangganan Deposito")
plt.xlabel("Status Deposito")
plt.ylabel("Jumlah Nasabah")

plt.show()
```

```python
# Hadijah Tohepaly - 202431078
# Visualisasi 2: Rata-rata Saldo Nasabah

saldo = df_clean.groupby("status_deposito")["balance"].mean()

saldo.plot(kind="bar", figsize=(8,5))

plt.title("Rata-rata Saldo Berdasarkan Status Deposito")
plt.xlabel("Status Deposito")
plt.ylabel("Rata-rata Saldo")

plt.show()
```

```python
# Hadijah Tohepaly - 202431078
# Visualisasi 3: Deposito Berdasarkan Pekerjaan

plt.figure(figsize=(12,6))

sns.countplot(
    data=df_clean,
    x="job",
    hue="status_deposito"
)

plt.title("Status Deposito Berdasarkan Pekerjaan")
plt.xlabel("Jenis Pekerjaan")
plt.ylabel("Jumlah Nasabah")
plt.xticks(rotation=45)

plt.show()
```
