```python
# Hadijah Tohepaly - 202431078
# Import Library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Hadijah Tohepaly - 202431078
# Load Dataset
df = pd.read_csv("bank-full.csv", sep=";")

print("Dataset berhasil dimuat")
print("Jumlah baris dan kolom:", df.shape)

df.head()
```

```python
# Hadijah Tohepaly - 202431078
# Informasi Dataset
df.info()
```

```python
# Hadijah Tohepaly - 202431078
# Mengecek Data Duplikat

print("Jumlah data duplikat:", df.duplicated().sum())
```

```python
# Hadijah Tohepaly - 202431078
# Mengecek Nilai Unknown

for col in df.select_dtypes(include="object").columns:
    print(f"\nKolom {col}:")
    print(df[col].value_counts())
```

```python
# Hadijah Tohepaly - 202431078
# ETL - Transform

# Membuat salinan dataset
df_clean = df.copy()

# Mengubah target deposito menjadi numerik
df_clean["deposit"] = df_clean["y"].map({
    "yes": 1,
    "no": 0
})

# Membuat kategori umur nasabah
df_clean["age_group"] = pd.cut(
    df_clean["age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["<=25", "26-35", "36-45", "46-55", ">55"]
)

# Mengubah nama kolom y menjadi status_deposito
df_clean = df_clean.rename(
    columns={"y": "status_deposito"}
)

print("Transform berhasil dilakukan")
df_clean.head()
```

```python
# Hadijah Tohepaly - 202431078
# Menyimpan hasil ETL
df_clean.to_csv("bank_cleaned.csv", index=False)

print("Data hasil ETL berhasil disimpan")
print("Jumlah data:", df_clean.shape)
```
